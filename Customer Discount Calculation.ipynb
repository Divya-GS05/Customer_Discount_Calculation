```python
Customers = {}
for i in range(5):
  Name = input("Enter the Customer Name: ")
  Purchased_Amount = int(input(f"Enter the Purchased amount of the customer {Name}: "))
  Customers[Name] = Purchased_Amount
print("\nCustomer Details: ",Customers)
for name,amount in Customers.items():
  print("\nCustomer Name: ", name)
  print("Purchased Amount: ", amount)
  if amount > 5000:
    print("Dicount = 1000/-")
    amount = amount - 1000
    print("Final Amount: ", amount)
  elif amount > 2000 and amount <= 5000:
    print("Discount = 500/-")
    amount = amount - 500
    print("Final Amount: ", amount)
  else:
    print("Discount = 0/-")
    print("Final Amount: ", amount)



```
